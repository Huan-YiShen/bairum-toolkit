```python
import matplotlib.pyplot as plt
import numpy as np

from plot_energey_transitions.plot_493 import plot_493_lab, plot_493_theory
## All numerical values are in unit (THz)
```

# 493 nm - setup inputs & functions

```python
'''Input values - use dummy data'''
### freq of Ba 137 and 138 transition
lab_ba138_freq493 = 607.4233 + 0.002550 + 0.0002 # wavemeter + EOM + AOM

### EOM freq
lab_central_freq_137 = 607.42765 + 0.000200 # lab value recorded on 20260603: Trapping Ba137+ 
ba137_EOM493_12 = 0.0040185 # EOM ch1 (carrier + ba137_EOM493_12)
ba137_EOM493_22 = -0.0040185 # EOM ch2 (carrier + ba137_EOM493_22)
```

# verify input value is sensible

```python
'''
493 relative theoritical calculation - with our Ba138 frequency as the anchor (absolute reference)
'''
# Relative theory value - absolute reference is on out 138 value
plt.plot() # Sample data
plot_493_theory(lab_ba138_freq493, "our_ba138_SP (lab value)", plotCentroid=True)
plt.xlim(607.421, 607.436)
plt.ylim(-0.1, 1.1)
plt.xlabel("Frequency (THz)")
plt.title ("493 nm transition for 138Ba and 137Ba ")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Plot lab 137
plot_493_theory(lab_ba138_freq493, "our_ba138_SP (lab)")
plot_493_lab(lab_central_freq_137, "central frequency (lab)", ba137_EOM493_12, ba137_EOM493_22, lab_ba138_freq493, plot_not_used = False)

plt.xlim(607.421, 607.436)
plt.ylim(-0.1, 1.1)
plt.xlabel("Frequency (THz)")
plt.title ("493 nm transition for 138Ba and 137Ba ")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
