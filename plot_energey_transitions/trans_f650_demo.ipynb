```python
import matplotlib.pyplot as plt
import numpy as np

from plot_650 import *
## All numerical values are in unit (THz)
```

```python
'''Input values - use dummy data'''
# input values
our_ba138_freq650 = 461.311_725 # lab value what the ion sees
```

# Theory value using an absolute lab reference

```python
''' Theory calculation'''
plot_650_theory(our_ba138_freq650, "our_ba138_freq650 (lab value)")

plt.xlim(461.310, 461.312)
plt.ylim(-0.1, 1.1)
plt.xlabel("Frequency (THz)")
plt.title ("138Ba and 137Ba 650 nm transition - theory")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

# Experimental value

```python
ba137_freq650_carrier_n_AOM = (461.31102 + 0.00059) + 0.0002 # WM + AOM
ba137_EOM650_D0P1 = -0.0003825
ba137_EOM650_D1P1 = -0.000525
ba137_EOM650_D2P1 = -0.000885
plot_650_theory(our_ba138_freq650, ref_name = "ba138_freq493_ref")
plot_650_lab_carD3P2(ba137_freq650_carrier_n_AOM, 
             ba137_EOM650_D0P1 = ba137_EOM650_D0P1, 
             ba137_EOM650_D1P1 = ba137_EOM650_D1P1,
             ba137_EOM650_D2P1 = ba137_EOM650_D2P1,
             our_ba138_freq650 = our_ba138_freq650,
             ref_name="ba137_freq650_carrier", data_name = "our")

plt.xlim(461.31102-0.001, 461.31102+0.001)
plt.ylim(-0.1, 1.1)
plt.xlabel("Frequency (THz)")
plt.title ("650 nm transition for 138Ba and 137Ba ")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
