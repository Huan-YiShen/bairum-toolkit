```python
import matplotlib.pyplot as plt
import numpy as np

from plot_energey_transitions.plot_650 import plot_650_lab, plot_650_theory
## All numerical values are in unit (THz)
```

```python
'''Input values - use dummy data'''
# input values
our_ba138_freq650 = 461.311_725 # lab value what the ion sees
```

# Theory value using an absolute lab reference

```python
''' Theory calculation'''
plot_650_theory(our_ba138_freq650, "our_ba138_freq650 (lab value)")

plt.xlim(461.310, 461.312)
plt.ylim(-0.1, 1.1)
plt.xlabel("Frequency (THz)")
plt.title ("138Ba and 137Ba 650 nm transition - theory")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

# Experimental value

```python
lab_ba138_freq650_central = 461.31182 # THz - lab value of what the ion sees
# EOM settings
ba137_EOM650_D2P1 = -0.000_885
ba137_EOM650_D1P1 = -0.000_525
ba137_EOM650_D0P1 = -0.000_382

plt.plot()
plot_650_lab(lab_ba138_freq650_central, "lab_ba138_freq650_central = D3P2",
            ba137_EOM650_D0P1 = ba137_EOM650_D0P1,
            ba137_EOM650_D1P1 = ba137_EOM650_D1P1,
            ba137_EOM650_D2P1 = ba137_EOM650_D2P1,
            our_ba138_freq650 = our_ba138_freq650,
            plot_not_used = True)
plt.xlim(461.310, 461.312)
plt.ylim(-0.1, 1.1)
plt.xlabel("Frequency (THz)")
plt.title ("138Ba and 137Ba 650 nm transition")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
